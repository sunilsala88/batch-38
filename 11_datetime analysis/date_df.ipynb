```python
import pandas as pd, requests, io

url = "https://en.wikipedia.org/wiki/NIFTY_50"
html = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}).text

tables = pd.read_html(io.StringIO(html))   # 33 tables on the page
df = tables[1].dropna(how="all")           # constituents list
df.columns = ["company", "symbol", "sector", "date_added"]
df
```

```python
df["date_added"] = pd.to_datetime(
    df["date_added"]
      .str.replace(r"\[.*?\]", "", regex=True)   # drop footnote markers [a], [h], ...
      .str.replace("\xa0", " ")                  # non-breaking spaces from Wikipedia
      .str.strip(),
    format="%d %B %Y",
)
```
