```python
dt2=dt2+dt.timedelta(days=1)
print(dt2)
```

```python
thurdays=[]
start=dt.datetime(2025, 1, 1)

for i in range(365):
    if start.weekday()==3:
        thurdays.append(start)

    start=start+dt.timedelta(days=1)
```
