```python
#circle class
#instanace attributes: radius
#class attributes: pi=3.14
#2 methods: area , circumference
```

```python
class Circle:
    pi=3.14  # class attribute

    def __init__(self,radius):
        self.radius=radius  # instance attribute

    def __repr__(self):
        return f"Circle(radius={self.radius})"

    def area(self):
        return self.pi * self.radius * self.radius

    def circumference(self):
        return 2 * self.pi * self.radius
    

c1=Circle(5)
print(c1.area())

c2=Circle(10)
print(c2.circumference())
```

```python
print(c1.area())
print(Circle.area(c1))  # calling area method using class name and passing instance as argument
```

```python

class Book:
    def __init__(self,title,author):
        self.title=title
        self.author=author
        self.is_available=True

    def borrow(self):
        self.is_available=False
        return f"{self.title} by {self.author} borrowed successfully."

    def return_book(self):
        self.is_available=True
        return f"{self.title} by {self.author} returned successfully."
```

```python
book1 = Book("Harry Potter", "J.K. Rowling")
print(book1.borrow())
print(book1.return_book())
```

```python
class BankAccount:
    bank_name="ABC Bank"  # class attribute

    def __init__(self,name,balance):
        self.name=name  # instance attribute
        self.balance=balance  # instance attribute

    def deposit(self,amount):
        self.balance=self.balance + amount
        print(f"Deposit of {amount} successful.")

    def withdraw(self,amount):
        if amount > self.balance:
            print("Insufficient balance.")
        else:
            self.balance=self.balance - amount
            print(f"Withdrawal of {amount} successful.")

    def check_balance(self):
        print(f"Account holder: {self.name}, Balance: {self.balance}")
```

```python
c1=BankAccount("Alice",1000)
c1.deposit(500)
c1.withdraw(200)
c1.check_balance()
```

```python
class Trading:
    stock_prices={"AAPL":150,"GOOGL":2800,"AMZN":3400}  # class attribute

    def __init__(self,name,balance):
        self.name=name  # instance attribute
        self.balance=balance  # instance attribute
        self.portfolio={}  # instance attribute

    def buy_stock(self,stock,quantity):
        if stock in self.stock_prices:
            total_cost=self.stock_prices.get(stock,0)*quantity
            if total_cost > self.balance:
                print("Insufficient balance to buy stocks.")
            else:
                self.balance=self.balance - total_cost
                if stock in self.portfolio:
                    
                    current_quantity=self.portfolio.get(stock)
                    self.portfolio.update({stock: current_quantity + quantity})
                else:
                    self.portfolio.upadate({stock:quantity})
                print(f"Bought {quantity} shares of {stock}.")
        else:
            print("Stock not available.")

    def sell_stock(self,stock,quantity):
        if stock in self.portfolio and self.portfolio[stock]>=quantity:
            total_earnings=self.stock_prices[stock]*quantity
            self.balance=self.balance + total_earnings
            self.portfolio[stock]-=quantity
            if self.portfolio[stock]==0:
                del self.portfolio[stock]
            print(f"Sold {quantity} shares of {stock}.")
        else:
            print("Insufficient shares to sell.")

    def check_portfolio(self):
        print(f"Account holder: {self.name}, Balance: {self.balance}, Portfolio: {self.portfolio}")
```
