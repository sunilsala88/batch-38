```python
import requests, re, pandas as pd

s = requests.Session()
url = "https://chartink.com/screener/nifty-50-stocks"
h = {"User-Agent": "Mozilla/5.0"}

page = s.get(url, headers=h, timeout=30)
csrf = re.search(r'name="csrf-token" content="([^"]+)"', page.text).group(1)

h |= {"x-csrf-token": csrf, "x-requested-with": "XMLHttpRequest", "referer": url}
resp = s.post("https://chartink.com/screener/process",
              headers=h,
              data={"scan_clause": "( {33489} ( latest close > 0 ) )"},
              timeout=30)

df = pd.DataFrame(resp.json()["data"])
df
#request,selelium,playwright,beautifulsoup,requests,seleniumwire,undetected_chromedriver,pyppeteer,pyppeteer_stealth
#antigravity
```
