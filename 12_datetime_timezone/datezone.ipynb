```python
def main():
    print('running strategy')

current_time=dt.datetime.now()
start_time=dt.datetime(current_time.year,current_time.month,current_time.day,18,34,0)
end_time=dt.datetime(current_time.year,current_time.month,current_time.day,18,35,0)

while True:
    current_time=dt.datetime.now()
    print(current_time)
    if current_time>=start_time and current_time<=end_time:
        if current_time.second%5==0:
            main()
    if current_time>end_time:
        break
    time.sleep(1)
print('strategy completed')
```
